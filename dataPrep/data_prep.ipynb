```python
import pandas as pd
data = pd.read_csv('Urdu/urdu_progress.csv')
print(data.head(20))
print(len(data))
```

```python
def map_stance(row):
    c = row['stance_confidence']
    label = row['stance_label']
    if c < 0.5 or ( 0.7 < c < 0.8):
        return None  # discard
    if label == "entailment" and c > 0.8:
        return "agree"
    if label == "contradiction" and c > 0.8:
        return "disagree"
    if label == "neutral":
        if 0.5 <= c <= 0.6:
            return "discuss"
        elif 0.6 < c <= 0.8:
            return "discuss"  # optional: include some medium confidence
    if c >= 0.5 and c <= 0.6:
        return "unrelated"
    return "unrelated"  # fallback for medium confidence


def map_sentiment(row):
    c = row['sentiment_confidence']
    label = row['sentiment_label']
    if c < 0.5 or ( 0.7 < c < 0.8):
        return None  # discard
    if label.lower() == "positive" and c > 0.8:
        return "positive"
    if label.lower() == "negative" and c > 0.8:
        return "negative"
    if 0.5 <= c <= 0.7:
        return "neutral"  # include medium confidence
    return "neutral"  # fallback

data['stance_new'] = data.apply(map_stance, axis=1)
data['sentiment_new'] = data.apply(map_sentiment, axis=1)

# Optionally, drop discarded rows
data_reduced = data.dropna(subset=['stance_new', 'sentiment_new'])
```

```python
print(data_reduced.head(20))
```

```python
print(len(data_reduced))
print(data_reduced['stance_new'].value_counts())
print(data_reduced['sentiment_new'].value_counts())
```

```python
# Separate neutral/unrelated vs others
neutral_df = data_reduced[data_reduced['stance_new'] == 'unrelated']
other_df = data_reduced[data_reduced['stance_new'] != 'unrelated']

# Decide target neutral size (e.g., reduce by 20-30%)
target_neutral = int(len(neutral_df) * 0.55)  # keeps 70% of neutral

neutral_sampled = neutral_df.sample(n=target_neutral, random_state=42)

# Combine with other classes
data_reduce = pd.concat([neutral_sampled, other_df]).sample(frac=1, random_state=42)  # shuffle

print(data_reduce['stance_new'].value_counts())
```

```python
print(len(data_reduce))
print(data_reduce['stance_new'].value_counts())
print(data_reduce['sentiment_new'].value_counts())
```

```python
data_reduce = data_reduce.sample(frac=1, random_state=42).reset_index(drop=True)
```

```python
data_reduce = data_reduce[['Headline', 'News Text', 'stance_new', 'sentiment_new']]
```

```python
import contractions
from cleantext import clean

def normalize_text(text):
    if not isinstance(text, str):
        return ""  # handle missing/NaN values
    # Step 1: expand contractions
    text = contractions.fix(text)
    # Step 2: clean the text
    text = clean(
        text,
        fix_unicode=True,
        to_ascii=True,
        lower=False,
        no_urls=True,
        no_emails=True,
        no_numbers=False,
        no_punct=False,
        replace_with_punct=""
    )
    return text
```

```python
# Apply to 'headline' column
eng['headline_clean'] = eng['Headline'].apply(normalize_text)

# Apply to 'body' column
eng['body_clean'] = eng['News Text'].apply(normalize_text)
```

```python
import re
def normalize_urdu(text):
    if not isinstance(text, str):
        return ""
    
    # Remove URLs
    text = re.sub(r"http\S+|www\S+|https\S+", "", text)
    
    # Remove emails
    text = re.sub(r"\S+@\S+", "", text)
    
    # Remove special invisible characters
    text = text.replace("\u200C", " ")  # ZWNJ
    text = text.replace("\u200F", " ")  # RLM
    text = text.replace("ٖ", "")        # diacritics
    
    # Keep letters, digits, whitespace, Urdu punctuation
    text = re.sub(r"[^\w\sء-ۿ،؟۔]", "", text)
    
    # Replace multiple spaces with single space
    text = re.sub(r"\s+", " ", text).strip()
    
    return text
```

```python
# Apply to 'headline' column
urdu['headline_clean'] = urdu['Headline'].apply(normalize_urdu)

# Apply to 'body' column
urdu['body_clean'] = urdu['News Text'].apply(normalize_urdu)
```
